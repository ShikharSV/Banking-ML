```python
 from datasets import load_dataset
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, f1_score

ds = load_dataset("PolyAI/banking77")
X_train, y_train = ds["train"]["text"], ds["train"]["label"]
X_test, y_test = ds["test"]["text"], ds["test"]["label"]

vec = TfidfVectorizer(ngram_range=(1, 2), sublinear_tf=True)
Xtr = vec.fit_transform(X_train)
Xte = vec.transform(X_test)

for name, model in [("LogReg", LogisticRegression(max_iter=1000)), ("LinearSVC", LinearSVC())]:
    model.fit(Xtr, y_train)
    pred = model.predict(Xte)
    train_acc = accuracy_score(y_train, model.predict(Xtr))
    print(name,
          "| TEST accuracy:", round(accuracy_score(y_test, pred), 4),
          "| TEST macro-F1:", round(f1_score(y_test, pred, average="macro"), 4),
          "| (train accuracy:", round(train_acc, 4), ")")
```
